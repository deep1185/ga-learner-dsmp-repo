```python
df = pd.DataFrame(ws_calc_records)
df.head()
```

```python
df['Forecast'] = df['NRF Landing']*df['PPC']
df.head()
```

```python
import numpy as np
# create a list of our conditions
conditions = [
    (df['PNA cases'] == 0)&(df['Forecast']== 0),
    (df['PNA cases'] == 0)&(df['Forecast']!= 0),
    (df['PNA cases']!= 0)&(df['Forecast']== 0),
    (df['PNA cases']!= 0)&(df['Forecast']!= 0)]

# create a list of the values we want to assign for each condition
values = [df['Forecast'],df['Forecast'],1,df['Forecast']]

# create a new column and use np.select to assign values to it using our lists as arguments
df['New Forecast'] = np.select(conditions, values)
df.head(100)
```

```python
df['New Forecast'].unique()
```

```python
df['New Forecast'].sum()
```

```python
import numpy as np
import math
df['Final Forecast']=df['New Forecast'].apply(lambda X: int(math.ceil(X)))
df.head()
```

```python
df['Final Forecast'].sum()
```

```python
pivot=df.groupby(['part code','Price','Prod Cat','PC Brand']).agg({'Final Forecast':'sum'})
forecast_pivot = pd.DataFrame(pivot)
forecast_pivot
```

```python
forecast_pivot['Final Forecast'].sum()
```
